```python
from pathlib import Path
import subprocess
import sys

repo = Path('/content/Fooocus')
if repo.exists():
    # Keep an existing checkout current; a failed pull (e.g. local edits) leaves it as is.
    subprocess.run(['git', '-C', str(repo), 'pull', '--ff-only'])
else:
    subprocess.run(['git', 'clone', 'https://github.com/b45tet/Fooocus.git', str(repo)], check=True)
if not (repo / 'colab.py').is_file():
    raise RuntimeError('This Fooocus checkout predates the Colab launcher. Update it or use a fresh runtime.')

command = [sys.executable, '-u', 'colab.py', '--share', '--always-high-vram',
           '--always-download-new-model', '--preset', 'b45tet']
with subprocess.Popen(command, cwd=repo, stdout=subprocess.PIPE,
                      stderr=subprocess.STDOUT, text=True, errors='replace') as process:
    for line in process.stdout:
        print(line, end='', flush=True)
    if process.wait():
        raise RuntimeError(f'Fooocus exited with status {process.returncode}. See the launcher output above.')
```
